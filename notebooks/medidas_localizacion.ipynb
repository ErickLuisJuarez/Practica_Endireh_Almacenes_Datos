### Media y mediana ponderadas por `factor_expansion`

La ENDIREH usa muestreo complejo: cada registro representa a un número
distinto de mujeres en la población real. Reportar solo la media/mediana
simple describiría únicamente a la muestra encuestada, no a la población
nacional.


```python
media_pond_edad = calcular_media_ponderada(df, "edad_primer_union")
mediana_pond_edad = calcular_mediana_ponderada(df, "edad_primer_union")
media_pond_hijos = calcular_media_ponderada(df, "num_hijos")
mediana_pond_hijos = calcular_mediana_ponderada(df, "num_hijos")

print(f"edad_primer_union -> media ponderada: {media_pond_edad:.2f} | mediana ponderada: {mediana_pond_edad:.2f}")
print(f"num_hijos -> media ponderada: {media_pond_hijos:.2f} | mediana ponderada: {mediana_pond_hijos:.2f}")
```

**Interpretación:** se reportan ambas versiones (simple y ponderada) porque
difieren según cómo esté distribuido el factor de expansión entre entidades y
estratos. La versión ponderada es la que debe citarse para hablar
de la población nacional de mujeres de 15 años y más.
